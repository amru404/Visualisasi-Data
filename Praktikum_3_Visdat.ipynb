```python
import matplotlib.pyplot as plt
```

```python

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
product_A_sales = [10,20,15,25,30,45,40,50,60,55,65,70]
product_B_sales = [5,10,8,15,18,20,22,30,25,46,40,45]
```

```python
plt.plot(months, product_A_sales, label ='Product A')
plt.title("Penjualan Produk A per Bulan")
plt.xlabel('bulan')
plt.ylabel('penjualan')
plt.legend()
plt.grid(True)
plt.show()
```

```python
from matplotlib.lines import lineStyles
plt.plot(months, product_A_sales, label ='Product A', linestyle="--", marker='o')
plt.plot(months, product_B_sales, label ='Product B', linestyle="-", marker='x')

plt.title("Penjualan Produk per Bulan")
plt.xlabel('bulan')
plt.ylabel('penjualan')
plt.legend()
plt.grid(True)
plt.show()
```

**Jenis Garis Untuk Menunjukan Tren**

```python
#data sample tambahan
product_C_sales = [18,22,25,28,32,38,42,45,48,52,56,60]
product_D_sales = [7,9,11,13,16,18,20,23,25,28,30,33]

plt.plot(months, product_A_sales, label ='Product A', linestyle="--",color='red')
plt.plot(months, product_B_sales, label ='Product B', linestyle="-",color='blue')
plt.plot(months, product_C_sales, label ='Product C', linestyle=":",color='green')
plt.plot(months, product_D_sales, label ='Product D', linestyle="-.",color='orange')

plt.title("Penjualan Produk per Bulan")
plt.xlabel('bulan')
plt.ylabel('penjualan')
plt.legend()
plt.grid(True)
plt.show()
```

```python
fig, axs = plt.subplots(2, 1, figsize=(10,8))

#plot utk C
axs[0].plot(months,product_C_sales, label = 'Product C', color='green', marker='d')
axs[0].set_title("Penjualan Produk per C Bulan")
axs[0].set_xlabel('bulan')
axs[0].set_ylabel('penjualan')
axs[0].legend()
axs[0].grid(True)

#plot utk D
axs[1].plot(months,product_D_sales, label = 'Product D', color='pink', marker='s')
axs[1].set_title("Penjualan Produk per D Bulan")
axs[1].set_xlabel('bulan')
axs[1].set_ylabel('penjualan')
axs[1].legend()
axs[1].grid(True)

plt.tight_layout()
plt.show()
```
