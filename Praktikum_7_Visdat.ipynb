**1. Import Library**

```python
import matplotlib.pyplot as plt
import numpy as np
```

**2. Datasaet**

```python
brands = ['Samsung','Iphone','Poconk','KinkFinix']
sales_2023 = [120,230,170,400]
sales_2024 = [150,200,180,450]
sales_2025 = [170,250,200,480]
```

**basic**

```python
y_pos = np.arange(len(brands))

fig1, ax1, = plt.subplots()

ax1.set_title('Horizontal bar chart - 2023')
ax1.set_xlabel('Jumlah Penjualan')
ax1.set_ylabel('Merk')
ax1.barh(brands, sales_2023)
plt.show()
```

```python
y_pos = np.arange(len(brands))

fig1, ax1, = plt.subplots()

ax1.set_title('Horizontal bar chart - 2024')
ax1.set_xlabel('Jumlah Penjualan')
ax1.set_ylabel('Merk')
ax1.barh(brands, sales_2024)
plt.show()

fig2, ax2, = plt.subplots()
ax2.set_title('Horizontal bar chart - 2024 & 2025')
ax2.set_xlabel('Jumlah Penjualan')
ax2.set_ylabel('Merk')

ax2.barh(brands, sales_2024, label='2024')
ax2.barh(brands, sales_2025, left=sales_2024, label='2025')
ax2.legend()
plt.show()
```

```python
y_pos = np.arange(len(brands))
colors = ['gold', 'skyblue','red','coral']

fig3, ax3, = plt.subplots()
ax3.barh(brands, sales_2023, color=colors,edgecolor='black',height=.6, label='2023')
ax3.set_title('Horizontal bar chart - 2023')
ax3.set_xlabel('Jumlah Penjualan')
ax3.set_ylabel('Merk')
ax3.grid(axis='x', linestyle='--', alpha=0.7)

for i, v in enumerate(sales_2023):
  ax3.text(v + 10, i, str(v), color='black')

plt.show()

fig4, ax4, = plt.subplots()
ax4.barh(brands, sales_2023, color='lightcoral',edgecolor='black',height=.6, label='2023')
ax4.barh(brands, sales_2024, color='skyblue',left=sales_2023,edgecolor='black',height=.6, label='2024')
ax4.set_title('Horizontal bar chart - 2023 & 2024')
ax4.set_xlabel('Jumlah Penjualan')
ax4.set_ylabel('Merk')
ax4.grid(axis='x', linestyle='--', alpha=0.5)




total_sales = np.array(sales_2023) + np.array(sales_2024)

for i, v in enumerate(total_sales):
  ax4.text(v + 10, i, str(v), va='center')

plt.show()
```

**bold text**

```python
y_pos = np.arange(len(brands))

fig5, ax5 = plt.subplots()

#bar 1
ax5.barh(y_pos - 0.2 ,  sales_2023, color='lightcoral',edgecolor='black',height=0.4, label='2023')
ax5.barh(y_pos + 0.2 ,  sales_2024, color='blue',edgecolor='black',height=0.4, label='2024')
ax5.set_yticks(y_pos, brands)
ax5.set_title('Penjualan smartphone tahun - 2023 & 2024')
ax5.set_xlabel('Jumlah Penjualan')
ax5.set_ylabel('Merk')
ax5.legend()
plt.show()

fig6, ax6 = plt.subplots()

#bar 1
ax6.barh(y_pos, sales_2023, color='salmon',height=0.4, label='2023')
ax6.barh(y_pos, sales_2024, left=sales_2023, color='lightblue',height=0.4, label='2024')
ax6.barh(y_pos, sales_2025, left=np.array(sales_2023) + np.array(sales_2024), color='lightgreen',height=0.4, label='2025')
ax6.set_yticks(y_pos, brands)
ax6.set_title('Penjualan smartphone tahun - 2023 vs 2024 vs 2025')
ax6.set_xlabel('Jumlah Penjualan')
ax6.set_ylabel('Merk')
ax6.legend()
plt.show()
```
