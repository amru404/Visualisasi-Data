**Praktikum Bar Chart**

```python
import matplotlib.pyplot as plt

#data
jurusan = [
    "Ilmu komputer",
    "Teknik Informatika",
    "Sistem Informasi",
    "Data Science"
]

jumlah_mahasiswa = [120,150,100,80]

plt.bar(jurusan,jumlah_mahasiswa, color=['blue','orange','green','red'])
plt.xlabel('Jurusan')
plt.ylabel('Jumlah Mahasiswa')
plt.title('Jumlah Mahasiswa Per Jurusan')
plt.show()
```

```python
jurusan = [
    "Ilmu komputer",
    "Teknik Informatika",
    "Sistem Informasi",
    "Data Science"
]

jumlah_mahasiswa = [120,150,100,80]

plt.bar(jurusan,jumlah_mahasiswa, color=['blue','orange','green','red'])
plt.xlabel('Jurusan')
plt.ylabel('Jumlah Mahasiswa')
plt.title('Jumlah Mahasiswa Per Jurusan')
plt.xticks(rotation=45)

for i, v in enumerate(jumlah_mahasiswa):
    plt.text(i, v + 2, str(v), ha='center')

plt.show()
```

```python
jurusan = [
    "Ilmu komputer",
    "Teknik Informatika",
    "Sistem Informasi",
    "Data Science"
]

jumlah_mahasiswa = [120,150,100,80]
jumlah_mahasiswa_2024 = [110,140,95,85]

x = range(len(jurusan))
width = 0.4
ax = plt.subplot()
plt.bar(x, jumlah_mahasiswa, width, label = '2024', color='orange')
plt.bar([p + width for p in x], jumlah_mahasiswa_2024, width=width, label='2025', color='skyblue')
plt.xlabel('Jurusan')
plt.ylabel('Jumlah Mahasiswa')
plt.title('Jumlah Mahasiswa Per Jurusan angkatan 2024 vs 2025')
plt.xticks(rotation=45)
plt.xticks([p + width/2 for p in x], jurusan)
plt.legend()
plt.show()
```

```python
jurusan = [
    "Ilmu komputer",
    "Teknik Informatika",
    "Sistem Informasi",
    "Data Science"
]
tahun = [
    '2019',
    '2020',
    '2021',
    '2022',
    '2023'
]
data = {
    'Ilmu komputer': [100, 110, 120, 130,140],
    'Teknik Informatika': [120, 120, 132, 145,160],
    'Sistem Informasi': [90, 95, 100, 105,110],
    'Data Science': [70, 75, 80, 85, 90],
}

width = 0.2
x = range(len(tahun))

plt.bar(x, data["Ilmu komputer"], width=width, label="ilmu komputer", color='blue')
plt.bar([p + width for p in x], data['Sistem Informasi'], width=width, label="Sistem Informasi", color="green")
plt.bar([p + width*2 for p in x], data['Teknik Informatika'], width=width, label="Teknik Informatika", color="orange")
plt.bar([p + width*3 for p in x], data['Data Science'], width=width, label="Data Science", color="red")

plt.title('Tren Jumlah Mahasiswa Per Jurusan (2019 vs 2023)')
plt.xlabel('Tahun')
plt.ylabel('Jumlah Mahasiswa')
plt.xticks([p+width for p in x], tahun)
plt.legend()
plt.show()
```
