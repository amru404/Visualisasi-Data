```python
import matplotlib.pyplot as plt
```

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun', 'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec']
product_A_sales = [10,20,15,25,30,45,40,50,60,55,65,70]
product_B_sales = [5,10,8,15,18,20,22,30,25,46,40,45]
```

```python
fig, axs = plt.subplots(2,1,figsize=(10,8))
#plot utk C
axs[0].plot(months,product_A_sales, label = 'Product A', color='green', marker='d')
axs[0].set_title("Penjualan Produk per A Bulan")
axs[0].set_xlabel('bulan')
axs[0].set_ylabel('penjualan')
axs[0].legend()
axs[0].grid(True)

#plot utk D
axs[1].plot(months,product_B_sales, label = 'Product B', color='pink', marker='s')
axs[1].set_title("Penjualan Produk per B Bulan")
axs[1].set_xlabel('bulan')
axs[1].set_ylabel('penjualan')
axs[1].legend()
axs[1].grid(True)

plt.tight_layout()
plt.show()
```
